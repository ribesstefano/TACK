```python
from datasets import load_dataset, DatasetDict
import pandas as pd

ds = load_dataset("ailab-bio/PROTAC-Degradation-Predictor-Dataset", "default")['train']
df = ds.to_pandas()
poi_seqs = df["POI_Sequence"].dropna().unique().tolist()
ligase_seqs = df["Ligase_Sequence"].dropna().unique().tolist()

print(f"Number of unique POI sequences: {len(poi_seqs)}")
print(f"Number of unique Ligase sequences: {len(ligase_seqs)}")
```

```python
import numpy as np

# Load the embeddings
emb = np.load("path/to/ESM-S/embeddings.npz", allow_pickle=True)
emb = {k: v for k, v in zip(emb["sequences"], emb["embeddings"])}
print(f"Number of embeddings/sequences: {len(emb)}")
print(f"Embedding dimension: {emb[poi_seqs[0]].shape}")
```

```python
from sklearn.decomposition import PCA

# Fit transform the POI sequences, so that 90% of variance is retained, and get
# the number of components
poi_embeddings = [emb[seq] for seq in poi_seqs if seq in emb]
pca_poi = PCA(n_components=0.9)
pca_poi.fit(np.vstack(poi_embeddings))
print(f"Number of PCA components for POI embeddings to retain 90% variance: {pca_poi.n_components_}")

# Fit transform the Ligase sequences, so that 90% of variance is retained, and
# get the number of components
ligase_embeddings = [emb[seq] for seq in ligase_seqs if seq in emb]
pca_ligase = PCA(n_components=0.9)
pca_ligase.fit(np.vstack(ligase_embeddings))
print(f"Number of PCA components for Ligase embeddings to retain 90% variance: {pca_ligase.n_components_}")
```
